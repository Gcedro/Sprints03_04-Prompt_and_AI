```python
schema_code = r'''
from typing import Optional
from pydantic import BaseModel, Field, field_validator


class ConsultaRecarga(BaseModel):
    carregador_id: Optional[str] = Field(
        default=None,
        description="Identificador do carregador"
    )

    intencao: str = Field(
        description="Intenção principal da consulta"
    )

    prioridade: Optional[str] = Field(
        default=None,
        description="Prioridade da consulta"
    )

    resposta: str = Field(
        description="Resposta ao usuário"
    )

    requer_profissional: bool = Field(
        default=False,
        description="Indica se a consulta requer profissional qualificado"
    )

    @field_validator("intencao")
    @classmethod
    def validar_intencao(cls, value):
        permitidas = {
            "status",
            "recarga",
            "energia",
            "sobrecarga",
            "priorizacao",
            "disponibilidade",
            "fora_escopo",
            "seguranca"
        }

        if value not in permitidas:
            raise ValueError(
                f"Intenção inválida. Use uma das opções: {permitidas}"
            )

        return value

    @field_validator("prioridade")
    @classmethod
    def validar_prioridade(cls, value):
        if value is None:
            return value

        permitidas = {"baixa", "media", "alta", "critica"}

        if value not in permitidas:
            raise ValueError(
                f"Prioridade inválida. Use: {permitidas}"
            )

        return value
'''

with open("src/schemas/consulta_ev.py", "w", encoding="utf-8") as f:
    f.write(schema_code)
```

```python
scope_code = r'''
import re


TERMOS_PERMITIDOS = [
    "goodwe",
    "carregador",
    "carregamento",
    "recarga",
    "veículo elétrico",
    "veiculo elétrico",
    "ve",
    "energia",
    "potência",
    "potencia",
    "kwh",
    "kw",
    "bateria",
    "eletroposto",
    "sobrecarga",
    "sessão",
    "sessao"
]


def validar_escopo(texto: str) -> bool:
    texto = texto.lower()

    return any(
        termo in texto
        for termo in TERMOS_PERMITIDOS
    )


def extrair_carregador(texto: str):
    match = re.search(
        r"\bGW-\d{2}\b",
        texto.upper()
    )

    if match:
        return match.group(0)

    return None
'''

with open("src/guardrails/scope_validator.py", "w", encoding="utf-8") as f:
    f.write(scope_code)
```

```python
moderation_code = r'''
JAILBREAK_TERMS = [
    "ignore as instruções",
    "ignore as instrucoes",
    "ignore todas as instruções",
    "ignore todas as instrucoes",
    "system prompt",
    "mostre seu prompt",
    "revele seu prompt",
    "jailbreak",
    "bypass",
    "ignore regras",
    "finja que você é",
    "finja que voce e",
    "desconsidere suas regras"
]


def detectar_prompt_injection(texto: str) -> bool:
    texto = texto.lower()

    return any(
        termo in texto
        for termo in JAILBREAK_TERMS
    )


def moderar(texto: str) -> tuple[bool, str]:
    if detectar_prompt_injection(texto):
        return (
            False,
            "Não posso ignorar minhas regras, revelar instruções internas "
            "ou executar tentativas de jailbreak."
        )

    return True, ""
'''

with open("src/guardrails/moderation.py", "w", encoding="utf-8") as f:
    f.write(moderation_code)
```

```python
memory_code = r'''
from typing import Dict

from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory


_STORE: Dict[str, InMemoryChatMessageHistory] = {}


def get_session_history(session_id: str):
    if session_id not in _STORE:
        _STORE[session_id] = InMemoryChatMessageHistory()

    return _STORE[session_id]


def adicionar_memoria(chain):
    return RunnableWithMessageHistory(
        chain,
        get_session_history,
        input_messages_key="input",
        history_messages_key="history",
    )


def limpar_memoria(session_id: str):
    _STORE.pop(session_id, None)
'''

with open("src/chain/memoria.py", "w", encoding="utf-8") as f:
    f.write(memory_code)
```

```python
builder_code = r'''
import os

from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_ollama import ChatOllama

from src.schemas.consulta_ev import ConsultaRecarga
from src.chain.memoria import adicionar_memoria


MODEL = os.getenv(
    "OLLAMA_MODEL",
    "gpt-oss:20b"
)

BASE_URL = os.getenv(
    "OLLAMA_BASE_URL",
    "http://localhost:11434"
)


def carregar_prompt():
    with open(
        "prompts/system_prompt_v2.md",
        "r",
        encoding="utf-8"
    ) as f:
        return f.read()


def criar_modelo(model_name=None):
    return ChatOllama(
        model=model_name or MODEL,
        base_url=BASE_URL,
        temperature=0.2,
        top_p=0.9,
        num_predict=500,
    )


def criar_chain(model_name=None):

    prompt = ChatPromptTemplate.from_messages([
        (
            "system",
            carregar_prompt()
        ),
        MessagesPlaceholder(
            variable_name="history"
        ),
        (
            "human",
            "{input}"
        )
    ])

    llm = criar_modelo(model_name)

    chain = prompt | llm

    return adicionar_memoria(chain)
'''

with open("src/chain/builder.py", "w", encoding="utf-8") as f:
    f.write(builder_code)
```

```python
context_code = r'''
import tiktoken


def contar_tokens(texto: str) -> int:
    encoder = tiktoken.get_encoding("cl100k_base")
    return len(encoder.encode(texto))


def medir_prompt(caminho="prompts/system_prompt_v2.md"):
    with open(caminho, "r", encoding="utf-8") as f:
        texto = f.read()

    return {
        "arquivo": caminho,
        "caracteres": len(texto),
        "tokens_estimados": contar_tokens(texto)
    }


if __name__ == "__main__":
    print(medir_prompt())
'''

with open("src/chain/context.py", "w", encoding="utf-8") as f:
    f.write(context_code)
```

```python
def montar_contexto_operacional():
    linhas = []

    for charger, dados in GOODWE_DATA.items():
        linhas.append(
            f"""
Carregador: {charger}
Status: {dados['status']}
Potência atual: {dados['potencia_atual']} kW
Limite: {dados['limite_potencia']} kW
Energia: {dados['energia_kwh']} kWh
Sessão: {dados['sessao']}
Usuário: {dados['usuario']}
Fonte: {dados['fonte']}
"""
        )

    return "\n".join(linhas)


print(montar_contexto_operacional())
```

```python
import time
import gradio as gr

from src.chain.builder import criar_chain
from src.guardrails.moderation import moderar
from src.guardrails.scope_validator import validar_escopo
from src.guardrails.scope_validator import extrair_carregador


chain = criar_chain()


def responder(message, history):

    inicio = time.perf_counter()

    permitido, mensagem = moderar(message)

    if not permitido:
        return mensagem

    if not validar_escopo(message):
        return (
            "Posso ajudar apenas com assuntos relacionados à "
            "GoodWe, veículos elétricos, carregamento e energia."
        )

    carregador = extrair_carregador(message)

    contexto = montar_contexto_operacional()

    pergunta = f"""
Contexto operacional:

{contexto}

Carregador identificado:
{carregador}

Pergunta do usuário:
{message}
"""

    try:
        resposta = chain.invoke(
            {
                "input": pergunta
            },
            config={
                "configurable": {
                    "session_id": "usuario_demo"
                }
            }
        )

        tempo = time.perf_counter() - inicio

        return resposta.content

    except Exception as e:
        return (
            "Não foi possível processar a consulta no momento. "
            f"Erro técnico: {str(e)}"
        )
```

```python
import json
import time
import sys
import os


from src.guardrails.moderation import moderar
from src.guardrails.scope_validator import validar_escopo


def avaliar(pergunta, esperado):

    inicio = time.perf_counter()

    permitido, moderacao = moderar(pergunta)

    if not permitido:
        resposta = moderacao

    elif not validar_escopo(pergunta):
        resposta = (
            "Posso ajudar apenas com assuntos relacionados "
            "à GoodWe, veículos elétricos, carregamento e energia."
        )

    else:
        resposta = "LLM_RESPONSE_AQUI"

    latencia = time.perf_counter() - inicio

    texto = resposta.lower()
    esperado_lower = esperado.lower()

    acertou = (
        esperado_lower in texto
        or (
            esperado_lower == "recusa"
            and not permitido
        )
        or (
            esperado_lower == "fora do escopo"
            and not validar_escopo(pergunta)
        )
    )

    return {
        "pergunta": pergunta,
        "resposta": resposta,
        "acertou": acertou,
        "latencia_segundos": round(latencia, 4)
    }


with open(
    "evals/eval_set.json",
    "r",
    encoding="utf-8"
) as f:
    dataset = json.load(f)


resultados = []

for item in dataset:
    resultados.append(
        avaliar(
            item["pergunta"],
            item["esperado"]
        )
    )


acuracia = (
    sum(r["acertou"] for r in resultados)
    / len(resultados)
)


saida = {
    "total": len(resultados),
    "acuracia": acuracia,
    "resultados": resultados
}


with open(
    "evals/sprint3_results.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        saida,
        f,
        ensure_ascii=False,
        indent=2
    )

print(json.dumps(saida, ensure_ascii=False, indent=2))
```
